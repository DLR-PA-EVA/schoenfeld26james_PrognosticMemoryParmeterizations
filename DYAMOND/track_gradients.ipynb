```python
import matplotlib.pyplot as plt

def plot_gradients(model_online):
    epochs = range(len(model_online.ODE_gradients))

    # Helper to extract metric
    def get(metric_list, key):
        return [d[key] for d in metric_list]

    # ODE gradients
    ode_mean = get(model_online.ODE_gradients, "mean")
    ode_p95  = get(model_online.ODE_gradients, "p95")
    ode_max  = get(model_online.ODE_gradients, "max")

    # NN gradients
    nn_mean = get(model_online.NN_gradients, "mean")
    nn_p95  = get(model_online.NN_gradients, "p95")
    nn_max  = get(model_online.NN_gradients, "max")

    plt.figure(figsize=(10, 6))

    # ODE
    plt.plot(epochs, ode_mean, label="ODE mean")
    plt.plot(epochs, ode_p95,  label="ODE p95")
    plt.plot(epochs, ode_max,  label="ODE max")

    # NN
    # plt.plot(epochs, nn_mean, label="NN mean")
    # plt.plot(epochs, nn_p95,  label="NN p95")
    # plt.plot(epochs, nn_max,  label="NN max")

    plt.yscale("log")
    plt.xlabel("Epoch")
    plt.ylabel("Gradient norm")
    plt.title("Gradient Norms (ODE + NN)")
    plt.legend()
    plt.tight_layout()
    plt.show()

plot_gradients(model_online)
```
